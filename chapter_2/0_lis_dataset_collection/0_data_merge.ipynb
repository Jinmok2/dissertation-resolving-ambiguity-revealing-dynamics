```python
## Import necessary libraries

import pandas as pd
import glob
import os
```

```python
## Check CSV files

# Get the list of CSV files

folder_path = './data'
csv_files = glob.glob(os.path.join(folder_path, '*.csv'))

# Print the total number of CSV files

print(f"Number of CSV files: {len(csv_files)} (Expected: 146)")
if len(csv_files) != 146:
    print("The number of files is different from expected. Please check.")

print("\nFile list (showing first 5):")
for file in csv_files[:5]:  # Print the first 5 files only
    print(f"- {os.path.basename(file)}")
```

```python
## Combine all CSV files into one large DataFrame

dataframes = []

for i, file in enumerate(csv_files):
    print(f"Processing: {i+1}/{len(csv_files)} - {os.path.basename(file)}")
    
    if i == 0:
        # Read the first file with header
        df = pd.read_csv(file)
    else:
        # For subsequent files, skip the first row (header) and read without header
        df = pd.read_csv(file, header=None, skiprows=1)
        df.columns = dataframes[0].columns[:len(df.columns)]  # Use column names from the first file and match columns safely

    # Add a column to indicate the source file name
    df['source_file'] = os.path.basename(file)
    dataframes.append(df)

# Concatenate all DataFrames into one large DataFrame

big_df = pd.concat(dataframes, ignore_index=True)
```

```python
## Display information about the combined DataFrame

print(f"\nCombined DataFrame number of rows: {big_df.shape[0]}")
print(f"Combined DataFrame number of columns: {big_df.shape[1]}")
big_df.head(10)
```

```python
### Code for OND


## Filter data

# Filter rows where 'Year' is between 2014 and 2024

filtered_df = big_df[(big_df['Year'] >= 2014) & (big_df['Year'] <= 2024)]


# Filter rows where 'Document Type' is 'Article' or 'Conference paper'

filtered_df = filtered_df[filtered_df['Document Type'].isin(['Article', 'Conference paper'])]


# Select necessary columns if they exist

columns_needed = ['Author full names', 'Author(s) ID', 'Title', 'Year', 'Source title', 'Affiliations', 'Funding Details', 
                  'Funding Texts', 'Language of Original Document', 'Document Type']
columns_found = [col for col in columns_needed if col in big_df.columns]

if columns_found:
    selected_df = filtered_df[columns_found]
    print(f"Created DataFrame with columns: {columns_found}")
else:
    print("None of the requested columns exist in the DataFrame.")


# Check the number of rows in the filtered DataFrame

print(f"\nNumber of rows after filtering: {selected_df.shape[0]}")


# Preview the first 10 rows of the filtered DataFrame
selected_df.head(10)
```

```python
## Write CSV file

selected_df.to_csv('selected_data.csv', index=False)
print("\nFiltered data saved as 'selected_data.csv'")
```
