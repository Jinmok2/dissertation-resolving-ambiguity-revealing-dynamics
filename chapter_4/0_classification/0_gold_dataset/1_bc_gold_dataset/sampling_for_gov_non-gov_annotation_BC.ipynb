```python
import pandas as pd

# =============================================================================
# 1. Load the CSV file
# =============================================================================
input_file = "deployment_final_ond_clusters.csv"
df = pd.read_csv(
    input_file,
    encoding="utf-8-sig"
)
```

```python
# =============================================================================
# 2. Remove duplicated organization names
# =============================================================================
unique_orgs = (
    df["Representative_Organization_Name"]
    .dropna()                     # Remove missing values
    .drop_duplicates()            # Keep only unique organization names
    .reset_index(drop=True)
)
```

```python
# =============================================================================
# 3. Check whether there are enough unique organizations
# =============================================================================
sample_size = 500

if len(unique_orgs) < sample_size:
    raise ValueError(
        f"Only {len(unique_orgs)} unique organization names are available, "
        f"which is fewer than the requested sample size ({sample_size})."
    )
```

```python
# =============================================================================
# 4. Randomly sample 500 unique organization names
# =============================================================================
sampled_orgs = unique_orgs.sample(
    n=sample_size,
    random_state=42
).reset_index(drop=True)
```

```python
# =============================================================================
# Save the sampled organization names to a CSV file
# =============================================================================
output_file = "sampled_representative_organizations_500_BC.csv"
sampled_orgs.to_frame(name="Representative_Organization_Name").to_csv(
    output_file,
    index=False,
    encoding="utf-8-sig"
)
```

```python
# =============================================================================
# Print summary information
# =============================================================================
print(f"Total rows in the original dataset: {len(df):,}")
print(f"Unique organization names: {len(unique_orgs):,}")
print(f"Randomly sampled organization names: {len(sampled_orgs):,}")
print(f"Output saved to: {output_file}")
```
