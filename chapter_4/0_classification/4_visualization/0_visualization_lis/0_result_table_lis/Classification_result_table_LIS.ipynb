```python
# =============================================================================
# Import libraries
# =============================================================================

import pandas as pd
from pathlib import Path
```

```python
# =============================================================================
# Define file paths
# =============================================================================

SVM_FILE = Path(
    "./svm_10fold_aggregate_metrics.csv"
)

BERT_FILE = Path(
    "./bert_10fold_aggregate_metrics.csv"
)

ZERO_SHOT_FILE = Path(
    "./zero_shot_10fold_aggregate_metrics.csv"
)

FEW_SHOT_FILE = Path(
    "./few_shot_10fold_aggregate_metrics.csv"
)
```

```python
# =============================================================================
# Load aggregate metric files
# =============================================================================

svm_df = pd.read_csv(
    SVM_FILE,
    encoding="utf-8-sig"
)

bert_df = pd.read_csv(
    BERT_FILE,
    encoding="utf-8-sig"
)

zero_shot_df = pd.read_csv(
    ZERO_SHOT_FILE,
    encoding="utf-8-sig"
)

few_shot_df = pd.read_csv(
    FEW_SHOT_FILE,
    encoding="utf-8-sig"
)
```

```python
# =============================================================================
# Define the metrics and their display labels
# =============================================================================

METRIC_LABELS = {
    "accuracy": "Accuracy",

    "class_0_precision": "Class 0 Precision",
    "class_0_recall": "Class 0 Recall",
    "class_0_f1": "Class 0 F1",

    "class_1_precision": "Class 1 Precision",
    "class_1_recall": "Class 1 Recall",
    "class_1_f1": "Class 1 F1",

    "macro_precision": "Macro Precision",
    "macro_recall": "Macro Recall",
    "macro_f1": "Macro F1",

    "weighted_precision": "Weighted Precision",
    "weighted_recall": "Weighted Recall",
    "weighted_f1": "Weighted F1"
}
```

```python
# =============================================================================
# Define a function to format mean and standard deviation
# =============================================================================

def format_mean_sd(
    dataframe,
    metric_name
):
    """
    Return the mean and standard deviation for a metric
    in the format: Mean (SD).
    """

    row = dataframe.loc[
        dataframe["Metric"] == metric_name
    ]

    if row.empty:
        return ""

    mean_value = row[
        "Mean"
    ].iloc[0]

    sd_value = row[
        "Standard_Deviation"
    ].iloc[0]

    return (
        f"{mean_value:.4f} "
        f"({sd_value:.4f})"
    )
```

```python
# =============================================================================
# Create the comparison table
# =============================================================================

table_rows = []

for metric_name, display_name in METRIC_LABELS.items():

    table_rows.append(
        {
            "Metric": display_name,

            "SVM": format_mean_sd(
                svm_df,
                metric_name
            ),

            "BERT": format_mean_sd(
                bert_df,
                metric_name
            ),

            "Llama Z-shot": format_mean_sd(
                zero_shot_df,
                metric_name
            ),

            "Llama F-shot": format_mean_sd(
                few_shot_df,
                metric_name
            )
        }
    )


comparison_table = pd.DataFrame(
    table_rows
)
```

```python
# =============================================================================
# Display the final comparison table
# =============================================================================

display(
    comparison_table
)
```

```python
# =============================================================================
# Save the comparison table
# =============================================================================

comparison_table.to_csv(
    "LIS_model_performance_comparison.csv",
    index=False,
    encoding="utf-8-sig"
)
```
